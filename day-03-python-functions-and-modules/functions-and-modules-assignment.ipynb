## Question 1 — Simple Interest Calculator (Default Arguments)

Write a function `calculate_simple_interest(principal, rate=5, time=1)` that calculates and
returns the simple interest using the formula:

```
interest = (principal * rate * time) / 100
```

- `rate` should default to `5` (percent) if not provided.
- `time` should default to `1` (year) if not provided.

**Sample calls and expected output:**
```python
calculate_simple_interest(1000, 10, 2)   # -> 200.0
calculate_simple_interest(1000)          # -> 50.0   (uses default rate=5, time=1)
calculate_simple_interest(2000, time=3)  # -> 300.0  (uses default rate=5)
```

````python
# Simple interest calculation

def calculate_simple_interest(principal, rate=5, time=1):
    return (principal * rate * time) / 100


# Function test
print(calculate_simple_interest(1000, 10, 2))
print(calculate_simple_interest(1000))
print(calculate_simple_interest(2000, time=3))
````

## Question 2 — Class Average Using `*args`

Write a function `class_average(*scores)` that accepts **any number of scores** and returns
their average, rounded to 2 decimal places.

- If no scores are passed at all, the function should return `0`.

**Sample calls and expected output:**
```python
class_average(80, 90, 70)         # -> 80.0
class_average(55, 60, 65, 70, 75) # -> 65.0
class_average()                   # -> 0
```

````python
# Class Average Code

def class_average(*scores):
    if not scores:
        return 0
    return round(sum(scores) / len(scores), 2)


# Function test
print(class_average(80, 90, 70))
print(class_average(55, 60, 65, 70, 75))
print(class_average(0))
````

## Question 3 — Analyze Numbers (Multiple Return Values + Built-ins)

Write a function `analyze_numbers(numbers)` that takes a list of numbers and returns **four
values** in this order:
1. the smallest number
2. the largest number
3. the sum of all numbers
4. the numbers sorted in descending order (as a list)

Use the built-in functions `min()`, `max()`, `sum()`, and `sorted()` — do not calculate these
manually with loops.

**Sample call and expected output:**
```python
smallest, largest, total, desc = analyze_numbers([4, 9, 1, 7, 3])
print(smallest)   # -> 1
print(largest)    # -> 9
print(total)      # -> 24
print(desc)       # -> [9, 7, 4, 3, 1]
```

````python
# Code to Analyze numbers

def analyze_numbers(numbers):
    return min(numbers), max(numbers), sum(numbers), sorted(numbers, reverse=True)


# Function test
smallest, largest, total, desc = analyze_numbers([4, 9, 1, 7, 3])
print(smallest)
print(largest)
print(total)
print(desc)
````

## Question 4 — Shared Booking Counter (Scope & the `global` keyword)

A small ticket-booking system needs to keep track of how many seats have been booked so far,
using a **global variable**.

1. Create a global variable `total_seats_booked` initialized to `0`.
2. Write a function `book_seats(n)` that:
   - adds `n` to `total_seats_booked` using the `global` keyword
   - prints a message like: `"Booked 3 seat(s). Total booked so far: 3"`
3. Write a second function `reset_bookings()` that resets `total_seats_booked` back to `0`
   (this also needs the `global` keyword).

**Sample calls and expected behaviour:**
```python
book_seats(3)     # Booked 3 seat(s). Total booked so far: 3
book_seats(5)     # Booked 5 seat(s). Total booked so far: 8
reset_bookings()
book_seats(2)     # Booked 2 seat(s). Total booked so far: 2
```

````python
# Code for Shared Booking Counter

total_seats_booked = 0

def book_seats(n):
    global total_seats_booked
    total_seats_booked += n
    print(f"Booked {n} seat(s). Total booked so far: {total_seats_booked}")

def reset_bookings():
    global total_seats_booked
    total_seats_booked = 0


# Function test
book_seats(3)
book_seats(5)
reset_bookings()
book_seats(2)
````

## Question 5 — Temperature Report Module (Custom Module + Standard Library)

This question has **two parts** and brings together functions, modules, and the standard
library.

**Part A — Build a module**

Create a module (a separate `.py` file, saved to disk) called `temperature_utils.py` containing:
- `celsius_to_fahrenheit(c)` → returns `(c * 9/5) + 32`
- `fahrenheit_to_celsius(f)` → returns `(f - 32) * 5/9`
- A constant `MODULE_VERSION = "1.0"`

**Part B — Use your module together with `random` and `datetime`**

Back in this notebook:
1. `import` your `temperature_utils` module.
2. Use the `random` module to generate 5 random Celsius temperatures between `15` and `40`
   (inclusive) — store them in a list.
3. Convert each one to Fahrenheit using your module's function, and store the results in a
   second list (a `for` loop or `map()` both work).
4. Use the `datetime` module to print today's date alongside the report, in the format:
   `"Temperature Report — DD-MM-YYYY"`
5. Print the Celsius list, the Fahrenheit list, and `temperature_utils.MODULE_VERSION`.

**Expected shape of the output (your actual numbers will differ, since they're random):**
```
Temperature Report — 11-08-2026
Celsius:    [22, 35, 18, 40, 27]
Fahrenheit: [71.6, 95.0, 64.4, 104.0, 80.6]
Module version: 1.0
```

````python
# Part A: write your module to disk first
module_code = '''
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

MODULE_VERSION = "1.0"
'''

with open("temperature_utils.py", "w") as f:
    f.write(module_code)

print("Module >> temperature_utils.py created !!")
````

````python
# Use module temparature_utils.py with 'random' and 'date time'

import random
import datetime
import temperature_utils

# 1. generate 5 random Celsius temperatures between 15 and 40 (inclusive)
celsius_temps = [random.randint(15, 40) for _ in range(5)]

# 2. convert each to Fahrenheit using your module
fahrenheit_temps = [temperature_utils.celsius_to_fahrenheit(c) for c in celsius_temps]

# 3. print today's date in the format: Temperature Report — DD-MM-YYYY
today = datetime.date.today()
print(f"Temperature Report — {today.strftime('%d-%m-%Y')}")

# 4. print the Celsius list, the Fahrenheit list, and the module version
print("Celsius:   ", celsius_temps)
print("Fahrenheit:", fahrenheit_temps)
print("Module version:", temperature_utils.MODULE_VERSION)
````
