```python
# Using
# loop
# continue
# isinstance()

# input
raw_values = [100, None, 250, "invalid", 300, None, 450, True]
valid_values = []

for values in raw_values:
    if not isinstance(values, int):
        continue
    if isinstance(values, bool):   # bool is a subclass of int, so exclude explicitly
        continue
    valid_values.append(values)

print(valid_values)
```

```python
# using list comprehension
raw_values = [100, None, 250, "invalid", 300, None, 450]

valid_values = [values for values in raw_values if isinstance(values, int)] #There are no True or False values in this list — thus, isinstance(values, bool) not required

print(valid_values)
```
